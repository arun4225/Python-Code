```python
# Different types of distribution

# Bernoulli Distribution
from scipy.stats import bernoulli
import seaborn as sns
ber_data = bernoulli.rvs(size=10000, p=0.6)

plot= sns.distplot(ber_data,
                 kde=False,
                 color="skyblue",
                 hist_kws={"linewidth": 15,'alpha':1})
plot.set(xlabel='Bernoulli Distribution', ylabel='Frequency')

# Uniform Distribution
from scipy.stats import uniform
import seaborn as sns
# random numbers from uniform distribution

uni_data = uniform.rvs(size=10000, loc = 10, scale=20)

uni_plot = sns.distplot(uni_data,
                  bins=100,
                  kde=True,
                  color='green',
                  hist_kws={"linewidth": 15,'alpha':1})
uni_plot.set(xlabel='Uniform Distribution ', ylabel='Frequency')

#Binomial Distribution
from scipy.stats import binom
import seaborn as sns
bino_data = binom.rvs(n=10,p=0.8,size=10000)

bino_plot = sns.distplot(bino_data,
                  kde=False,
                  color='green',
                  hist_kws={"linewidth": 15,'alpha':1})
bino_plot.set(xlabel='Binomial Distribution', ylabel='Frequency')

#Normal Distribution
from scipy.stats import norm
import seaborn as sns
norm_data = norm.rvs(size=10000,loc=0,scale=1)

norm_plot = sns.distplot(norm_data,
                  bins=100,
                  kde=True,
                  color='green',
                  hist_kws={"linewidth": 15,'alpha':1})
norm_plot.set(xlabel='Normal Distribution', ylabel='Frequency')

#Poisson Distribution
from scipy.stats import poisson
import seaborn as sns
pois_data = poisson.rvs(mu=3, size=10000)

pois_plot = sns.distplot(pois_data,
                  bins=30,
                  kde=False,
                  color='green',
                  hist_kws={"linewidth": 15,'alpha':1})
pois_plot.set(xlabel='Poisson Distribution', ylabel='Frequency')

#Exponential Distribution
from scipy.stats import expon
import seaborn as sns
expo_data = expon.rvs(scale=1,loc=0,size=1000)

expo_plot = sns.distplot(expo_data,
                  kde=True,
                  bins=100,
                  color='green',
                  hist_kws={"linewidth": 15,'alpha':1})
expo_plot.set(xlabel='Exponential Distribution', ylabel='Frequency')
```

```python
#Chi-square test
import scipy.stats as stats
import seaborn as sns
import pandas as pd
import numpy as np
dataset=sns.load_dataset('tips')
dataset.head()

dataset_table=pd.crosstab(dataset['sex'],dataset['smoker'])
print(dataset_table)

dataset_table.values
#Observed Values
Observed_Values = dataset_table.values 
print("Observed Values :-\n",Observed_Values)

val=stats.chi2_contingency(dataset_table)
val

Expected_Values=val[3]

no_of_rows=len(dataset_table.iloc[0:2,0])
no_of_columns=len(dataset_table.iloc[0,0:2])
ddof=(no_of_rows-1)*(no_of_columns-1)
print("Degree of Freedom:-",ddof)
alpha = 0.05

from scipy.stats import chi2
chi_square=sum([(o-e)**2./e for o,e in zip(Observed_Values,Expected_Values)])
chi_square_statistic=chi_square[0]+chi_square[1]


print("chi-square statistic:-",chi_square_statistic)
critical_value=chi2.ppf(q=1-alpha,df=ddof)
print('critical_value:',critical_value)

#p-value
p_value=1-chi2.cdf(x=chi_square_statistic,df=ddof)
print('p-value:',p_value)
print('Significance level: ',alpha)
print('Degree of Freedom: ',ddof)
print('p-value:',p_value)

if chi_square_statistic>=critical_value:
    print("Reject H0,There is a relationship between 2 categorical variables")
else:
    print("Retain H0,There is no relationship between 2 categorical variables")
    
if p_value<=alpha:
    print("Reject H0,There is a relationship between 2 categorical variables")
else:
    print("Retain H0,There is no relationship between 2 categorical variables")
```
