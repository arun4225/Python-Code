```python
df['quality'] = df['quality'].replace(8,7)
df['quality'] = df['quality'].replace(3,5)
df['quality'] = df['quality'].replace(4,5)
df['quality'].value_counts()
```

```python
X = df.drop('quality', axis=1)
y = df['quality']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=10)
```

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier,export_graphviz
from sklearn import metrics
from sklearn.tree import plot_tree
DT = DecisionTreeClassifier(criterion='entropy')
DT.fit(X_train, y_train)
DT.score(X_train, y_train)
DT.score(X_train, y_train)
DT.score(X_test, y_test)
```

```python
pred_test = DT.predict(X_test)
print(metrics.accuracy_score(y_test, pred_test))
```

```python
acc_DT = metrics.accuracy_score(y_test, pred_test)
resultDF = pd.DataFrame({'Method' : ['Decision Tree'], 'Accuracy' : acc_DT})
```

```python
clf_pruned = DecisionTreeClassifier(criterion='entropy', random_state=100, max_depth=3, min_samples_leaf=5)
clf_pruned.fit(X_train, y_train)
pred_pruned_train = clf_pruned.predict(X_train)
pred_pruned_test = clf_pruned.predict(X_test)
print(metrics.accuracy_score(y_test, pred_pruned_test))
print(metrics.accuracy_score(y_train, pred_pruned_train))
```

```python
#Apply Random Forest
from sklearn.ensemble import RandomForestClassifier
rfcl = RandomForestClassifier(n_estimators=50)
rfcl = rfcl.fit(X_train, y_train)
```

```python
pred_RF = rfcl.predict(X_test)
acc_RF = metrics.accuracy_score(y_test, pred_RF)
```

```python
tempResultDf = pd.DataFrame({'Method' : ['Random Forest'], 'Accuracy' : acc_RF})
resultDF = pd.concat([resultDF, tempResultDf])
```

```python
#Apply Adaboost
from sklearn.ensemble import AdaBoostClassifier
abcl = AdaBoostClassifier(n_estimators=100, learning_rate=0.1, random_state=10)
abcl = abcl.fit(X_train, y_train)
```

```python
pred_AB = abcl.predict(X_test)
acc_AB = metrics.accuracy_score(y_test, pred_AB)
```

```python
tempResultDf = pd.DataFrame({'Method' : ['AdaBoost'], 'Accuracy' : acc_AB})
resultDF = pd.concat([resultDF, tempResultDf])
```

```python
#Apply Bagging 
from sklearn.ensemble import BaggingClassifier
bgcl = BaggingClassifier(n_estimators=50, bootstrap=True, oob_score=True, random_state=10)
bgcl = bgcl.fit(X_train, y_train)
```

```python
pred_BG = bgcl.predict(X_test)
acc_BG = metrics.accuracy_score(y_test, pred_BG)
```

```python
tempResultDf = pd.DataFrame({'Method' : ['Boosting'], 'Accuracy' : acc_BG})
resultDF = pd.concat([resultDF, tempResultDf])
```
