```python
def convert_date(x):
    if x == '000000':
        return '1900-01-01'
    else:
        try:
            # Attempt to convert the date string to datetime
            return pd.to_datetime(x, format='%y%m%d').strftime('%Y-%m-%d')
        except ValueError:
            # Replace invalid dates with '1900-01-01'
            return '1900-01-01'

# Apply the function to the 'Defect_Card_Jic_Dt' column
df['Defect_Card_Jic_Dt'] = df['Defect_Card_Jic_Dt'].apply(convert_date)
```
