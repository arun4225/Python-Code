```python
class Student:
    def __init__(self, pName = 'NONAME', pRoll = 0):
        self.name = pName
        self.roll = pRoll
    
    def __del__(self):
        print('Deleting the student: ' + str(self))

    def __str__(self):
        vOutput = self.name + ' (' + str(self.roll) + ')'
        return vOutput

def main():
    objStudent1 = Student('Jack', 10)
    objStudent2 = Student('Jill', '11')
    objStudent3 = Student()   
    print('objStudent1: ', objStudent1) 
    print('objStudent2: ', objStudent2)
    print('objStudent3: ', objStudent3)

    del objStudent1

main()
```

```python
class Student:
    studentCount = 0

    def __int__(self, pName = 'NONAME', pRoll = 0, pMarks = 0.0):
        self.name = pName
        self.roll = pRoll
        self.marks = pMarks
        Student.studentCount += 1
        self.showCount()
    
    def __del__(self):
        print('Deleting the student: ' + str(self))
    
    def __str__(self):
        vOutput = self.name + ' (' + str(self.roll) + ')'
        return vOutput

    def showCount(self):
        print('Total student: ', Student.studentCount, str(self))
    
    def showStudentInfo(self):
        print('\nName: ', self.name)
        print('RollNo: ', self.roll)
        print('Marks: ', self.marks)
```
