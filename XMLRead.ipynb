```python
import xml.etree.ElementTree as ET

# Load and parse the XML file
file_path = '/content/sample_data/delek-pdi.xml'
tree = ET.parse(file_path)
root = tree.getroot()
print(root)

# Function to extract supplier information
def get_supplier_info(root):
    suppliers = []
    for header in root.findall('Header'):
        supplier = header.find('Supplier')
        if supplier is not None:
            supplier_info = {
                'VendorID': supplier.find('VendorID').text,
                'Name': supplier.find('Name').text,
                'Address': f"{supplier.find('Address1').text}, {supplier.find('Address2').text}, {supplier.find('City').text}, {supplier.find('State').text}, {supplier.find('Country').text}, {supplier.find('PostalCode').text}",
                'PhoneNumber': supplier.find('PhoneNumber').text
            }
            suppliers.append(supplier_info)
    return suppliers

# Extract and print supplier information
suppliers = get_supplier_info(root)
for supplier in suppliers:
    print(f"Supplier Name: {supplier['Name']}")
    print(f"Address: {supplier['Address']}")
    print(f"Phone Number: {supplier['PhoneNumber']}")
    print('')
```

```python
print(suppliers)
```

```python
def line_items(root):
    lines = []
    for header in root.findall('Header'):
        line = header.find('LineItems')
        if line is not None:
            line_items = {
                'ProductId' : line.find('ProductId').text,
                'ProductDescription' : line.find('ProductDescription').text,
                'BlendId' : line.find('BlendId').text,
                'InvoiceQuantity' : line.find('InvoiceQuantity').text,
                'Type' : line.findall('LineItemFreight').findall('Type').text,
                'QuantityBasisGrossorNet' : line.findall('LineItemFreight').findall('QuantityBasisGrossorNet').text,
                'Rate' : line.findall('LineItemFreight').findall('Rate').text,
                'Amount' : line.findall('LineItemFreight').findall('Amount').text
            }
            lines.append(line_items)
    return lines
```

```python
line = line_items(root)
print(line)
```
