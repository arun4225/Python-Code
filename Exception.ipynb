```python
def testFunction(pNumberToDivide):
    try:
        vNumber = int(input("Enter a number: "))
        vResult = pNumberToDivide / vNumber
        print("Result: ", vResult)
    except NameError:
        print("Please define variable before use!")
    except ZeroDivisionError:
        print("Division by zero is not possible")
    except:
        print("Unknown Error!")
    else:
        print("Congrats! No exception")
        print('The result is {}'.format(vResult))
    finally:
        print("Execution complete ..")
```

```python
testFunction(100)
```
